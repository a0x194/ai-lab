```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(0)
x = rng.uniform(0,3,30)
y = 2 * x + 1 + rng.normal(0,0.3,30)
w,b = 0.5,0.0
plt.scatter(x,y)
plt.plot(x,w * x + b,'r')
plt.show()
print(round(y.mean(),3))
```
